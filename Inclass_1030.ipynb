```python
class Games:
    num_of_games = 0
    
    
    def __init__(self,name = "None", price = 0, playtime = 0, playercount = 0, achievements = 0, multiplayer = False, controller=False, rating = "None", platform = "PC"):
        self.name = name
        self.price = price
        self.playtime = playtime
        self.playercount = playercount
        self.achievements = achievements
        self.multiplayer = multiplayer
        self.controller = controller
        self.rating = rating
        self.platform = platform
        Games.num_of_games +=1
        

    def add_playtime (self,minute):
        self.playtime = self.playtime + minute
        return 
        

    def achieved (self,num):
        self.achievements - num
        return 

    def discount(self,rate):
        self.price * rate
        return 

    def players(self,numPeople):
        self.playercount += numPeople
        return 

    def servers_down(self,Status):
        self.multiplayer = Status
        return 

    @staticmethod
    # Takes a list of games as input
    def most_played(games):
        mostPlayed = games[0]
        for i in games:
            if i.playtime > mostPlayed.playtime:
                mostPlayed = i
        return mostPlayed

class FPS(Games):
    pass

class RPG(Games):
    pass

class Sports(Games):
    pass

class Racing(Games):
    pass

class Survival(Games):
    pass

class Horror(Games):
    pass


minecraft = Games("Minecraft", 29.99, 0, 3000000, 20, True, True, "E10+", "PC")
game2 = Games("Roblox", 0.00, 0, 6000000,0, True, True, "E10+", "Everything")
    
        
```
